# Inspect the archived physics experiment
Run all cells to verify the packaged files and display the saved scientific results. No GPU or model training is used. Integrity labels describe file consistency; scientific labels describe the original evidence. Confidence intervals are not recomputed here.

Use a copy of this notebook in the same folder; saving executed outputs in the original changes its frozen release hash.


```python
from pathlib import Path
import subprocess, sys, html
from IPython.display import display, HTML
root = Path.cwd().resolve()
if not (root / "scripts/verify_release.py").is_file():
    raise RuntimeError("Open this notebook from the repository root.")
checked = subprocess.run([sys.executable, str(root / "scripts/verify_release.py")], cwd=root, text=True, capture_output=True)
ok = checked.returncode == 0
color = "#177245" if ok else "#b42318"
label = "INTEGRITY: GREEN" if ok else "INTEGRITY: RED — inspect diagnostics"
display(HTML(f'<p style="font-weight:bold;color:{color}">{html.escape(label)}</p>'))
print(checked.stdout)
if checked.stderr:
    print(checked.stderr)
```
